# Tokenization

文本一般被表示成 UTF-8 字符串，而语言模型在 token 序列上生成一个概率分布（通常用索引表示）。

Tokenizer 就是把 UTF-8 字符串 encode 成 token，把 token decode 成 UTF-8 字符串的工具。

[Interactive Tokenizer](https://tiktokenizer.vercel.app/?encoder=gpt2)

Observations
- Space Prefix: 单词和它前面的空格是同一个 token (e.g., " world").
- 同一个单词，位于开头和中间的时候是不同的 token (e.g., "hello hello").
- Number Chunking: 数字是分节的 (e.g., "189854" -> "189", "854").
- 中文多字合并: “你好”是一个单独的 token

```python
import tiktoken

tokenizer = tiktoken.get_encoding("o200k_base") # tokenizer for GPT-4o
string = "你好，世界！"

indices = tokenizer.encode(string)
reconstructed = tokenizer.decode(indices)
assert string == reconstructed    # tokenizer 应该是可逆的

for index in indices:
    print(f"{index}: {tokenizer.decode([index])}")
```
